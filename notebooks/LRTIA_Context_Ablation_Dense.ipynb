# LRTIA - Context Ablation (Dense Sampling)

**Dense sampling version** - especially fine-grained at short context lengths where we expect the steepest changes.

**Hypothesis:** Perplexity reduction follows diminishing returns:
- 0→16 tokens: HUGE drop (basic local context)
- 16→32 tokens: Large drop
- 32→64 tokens: Moderate drop
- 64→128 tokens: Small drop
- 128→256 tokens: Tiny drop (but intact > shuffled here)

The difference between intact and shuffled should emerge at longer contexts.

```python
@torch.no_grad()
def compute_perplexity_on_region(token_ids, target_start, target_end):
    """
    Compute perplexity only on tokens in [target_start, target_end].
    Context before target_start is provided but not scored.
    """
    if target_end > len(token_ids):
        target_end = len(token_ids)
    
    input_ids = torch.tensor([token_ids], device=model.device)
    
    outputs = model(input_ids)
    logits = outputs.logits[0]
    
    total_loss = 0.0
    count = 0
    
    for i in range(target_start, target_end - 1):
        log_probs = torch.log_softmax(logits[i], dim=-1)
        target_token = token_ids[i + 1]
        token_loss = -log_probs[target_token].item()
        total_loss += token_loss
        count += 1
    
    if count == 0:
        return float('inf'), 0
    
    avg_loss = total_loss / count
    perplexity = np.exp(avg_loss)
    
    return perplexity, avg_loss
```

```python
def analyze_context_dependency(text, context_lengths, target_size=40):
    """
    Measure perplexity on the last `target_size` tokens
    with varying amounts of preceding context.
    """
    full_tokens = tokenizer.encode(text)
    n_tokens = len(full_tokens)
    
    # Filter context lengths that are actually achievable
    max_possible_context = n_tokens - target_size
    valid_context_lengths = [c for c in context_lengths if c <= max_possible_context]
    
    if len(valid_context_lengths) < len(context_lengths):
        print(f"  Note: Text only supports up to {max_possible_context} tokens of context")
    
    target_start_in_full = n_tokens - target_size
    target_end_in_full = n_tokens
    
    results = []
    
    for ctx_len in valid_context_lengths:
        doc_start = max(0, target_start_in_full - ctx_len)
        truncated_tokens = full_tokens[doc_start:]
        
        target_start = len(truncated_tokens) - target_size
        target_end = len(truncated_tokens)
        
        ppl, loss = compute_perplexity_on_region(truncated_tokens, target_start, target_end)
        
        actual_context = target_start
        
        results.append({
            'requested_context': ctx_len,
            'actual_context': actual_context,
            'perplexity': ppl,
            'loss': loss,
        })
    
    return results
```

```python
# DENSE SAMPLING - fine-grained at short contexts
# Extended passages are ~500 tokens, max usable context is ~460
CONTEXT_LENGTHS = [
    # Very fine at low end (every 4 tokens)
    4, 8, 12, 16, 20, 24, 28, 32,
    # Medium granularity (every 8 tokens)
    40, 48, 56, 64,
    # Coarser (every 16 tokens)  
    80, 96, 112, 128,
    # Even coarser
    160, 192, 224, 256,
    # Max range
    320, 384, 448
]

TARGET_SIZE = 40  # Last 40 tokens are the target region
N_SHUFFLES = 5    # More shuffles for better averaging

print(f"Testing {len(CONTEXT_LENGTHS)} context lengths")
print(f"Range: {min(CONTEXT_LENGTHS)} to {max(CONTEXT_LENGTHS)} tokens")
print(f"Target region: last {TARGET_SIZE} tokens")
```

```python
all_results = []

passage_names = ['NN', 'Detective', 'Climate', 'Recipe', 'Rome']

for i, passage in enumerate(tqdm(PASSAGES, desc="Processing passages")):
    passage_name = passage_names[i]
    print(f"\n{passage_name}:")
    
    # INTACT
    intact_text = " ".join(passage.split())
    intact_results = analyze_context_dependency(intact_text, CONTEXT_LENGTHS, TARGET_SIZE)
    
    if intact_results:
        for r in intact_results:
            r['passage'] = passage_name
            r['condition'] = 'intact'
            all_results.append(r)
    
    # SHUFFLED (multiple versions)
    sentences = split_sentences(passage)
    for seed in range(N_SHUFFLES):
        rng = random.Random(42 + i * 100 + seed)
        shuffled = sentences.copy()
        rng.shuffle(shuffled)
        shuffled_text = " ".join(shuffled)
        
        shuffled_results = analyze_context_dependency(shuffled_text, CONTEXT_LENGTHS, TARGET_SIZE)
        
        if shuffled_results:
            for r in shuffled_results:
                r['passage'] = passage_name
                r['condition'] = 'shuffled'
                r['seed'] = seed
                all_results.append(r)

df = pd.DataFrame(all_results)
print(f"\n\nCollected {len(df)} measurements")
print(f"Unique context lengths tested: {sorted(df['actual_context'].unique())}")
```

```python
# Visualization with half-life markers
fig, ax = plt.subplots(figsize=(10, 6))

colors = {'intact': '#2ecc71', 'shuffled': '#e74c3c'}

# Compute half-lives for plotting
def get_half_life_point(cond_df):
    means = cond_df.groupby('actual_context')['perplexity'].mean().sort_index()
    contexts = means.index.values
    ppls = means.values
    
    total_benefit = ppls[0] - ppls[-1]
    target_ppl = ppls[0] - 0.5 * total_benefit
    
    for i in range(len(ppls) - 1):
        if ppls[i] >= target_ppl >= ppls[i+1]:
            frac = (ppls[i] - target_ppl) / (ppls[i] - ppls[i+1])
            hl_ctx = contexts[i] + frac * (contexts[i+1] - contexts[i])
            return hl_ctx, target_ppl
    return contexts[-1], ppls[-1]

for condition in ['intact', 'shuffled']:
    cond_df = df[df['condition'] == condition]
    means = cond_df.groupby('actual_context')['perplexity'].mean()
    stds = cond_df.groupby('actual_context')['perplexity'].std()
    sems = stds / np.sqrt(cond_df.groupby('actual_context').size())
    
    ax.errorbar(means.index, means.values, yerr=sems.values,
                marker='o', capsize=3, label=condition, 
                color=colors[condition], linewidth=2, markersize=5)
    
    # Mark half-life point
    hl_ctx, hl_ppl = get_half_life_point(cond_df)
    ax.axvline(hl_ctx, color=colors[condition], linestyle='--', alpha=0.5)
    ax.plot(hl_ctx, hl_ppl, marker='*', markersize=15, color=colors[condition], 
            markeredgecolor='black', markeredgewidth=1)
    ax.annotate(f'HL={hl_ctx:.0f}', (hl_ctx, hl_ppl), 
                textcoords="offset points", xytext=(10, 10 if condition=='intact' else -20),
                fontsize=10, color=colors[condition])

ax.set_xlabel('Context Length (tokens)', fontsize=12)
ax.set_ylabel('Perplexity', fontsize=12)
ax.set_title('Memory Curves with Half-Life Markers (★)', fontsize=14)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('context_ablation_with_halflife.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n★ = Half-life point (context length to reach 50% of total benefit)")
```

```python
# HALF-LIFE COMPUTATION
# The key scalar metric: context length to achieve 50% of total benefit

def compute_half_life(contexts, perplexities, percentile=0.5):
    """
    Compute the context length at which `percentile` of total benefit is achieved.
    Returns interpolated value for precision.
    """
    contexts = np.array(contexts)
    perplexities = np.array(perplexities)
    
    # Sort by context
    order = np.argsort(contexts)
    contexts = contexts[order]
    perplexities = perplexities[order]
    
    min_ppl = perplexities[0]   # Perplexity with minimal context
    max_ppl = perplexities[-1]  # Perplexity with maximal context
    total_benefit = min_ppl - max_ppl
    
    if total_benefit <= 0:
        return np.nan  # No improvement
    
    target_ppl = min_ppl - percentile * total_benefit
    
    # Find where we cross the target
    for i in range(len(perplexities) - 1):
        if perplexities[i] >= target_ppl >= perplexities[i+1]:
            # Linear interpolation
            frac = (perplexities[i] - target_ppl) / (perplexities[i] - perplexities[i+1])
            return contexts[i] + frac * (contexts[i+1] - contexts[i])
    
    return contexts[-1]  # Never reached target

print("=" * 70)
print("HALF-LIFE ANALYSIS (context tokens to reach 50% of benefit)")
print("=" * 70)

# Overall half-lives
half_lives = {}

for condition in ['intact', 'shuffled']:
    cond_df = df[df['condition'] == condition]
    means = cond_df.groupby('actual_context')['perplexity'].mean().sort_index()
    
    contexts = means.index.values
    perplexities = means.values
    
    hl_50 = compute_half_life(contexts, perplexities, 0.5)
    hl_25 = compute_half_life(contexts, perplexities, 0.25)
    hl_75 = compute_half_life(contexts, perplexities, 0.75)
    hl_90 = compute_half_life(contexts, perplexities, 0.90)
    
    half_lives[condition] = hl_50
    
    total_benefit = perplexities[0] - perplexities[-1]
    
    print(f"\n{condition.upper()}:")
    print(f"  Total perplexity reduction: {total_benefit:.2f}")
    print(f"  25% of benefit at: {hl_25:.0f} tokens")
    print(f"  50% of benefit at: {hl_50:.0f} tokens  <-- HALF-LIFE")
    print(f"  75% of benefit at: {hl_75:.0f} tokens")
    print(f"  90% of benefit at: {hl_90:.0f} tokens")

print(f"\n" + "-" * 50)
print(f"HALF-LIFE SUMMARY:")
print(f"  Intact:   {half_lives['intact']:.0f} tokens")
print(f"  Shuffled: {half_lives['shuffled']:.0f} tokens")
ratio = half_lives['shuffled'] / half_lives['intact']
print(f"  Ratio:    {ratio:.2f}x")
print(f"\n  Shuffled needs {ratio:.1f}x more context to get the same benefit.")

# Per-passage half-lives
print(f"\n" + "=" * 70)
print("PER-PASSAGE HALF-LIVES")
print("=" * 70)

passage_half_lives = []
for passage in df['passage'].unique():
    pass_df = df[df['passage'] == passage]
    
    row = {'passage': passage}
    for condition in ['intact', 'shuffled']:
        cond_df = pass_df[pass_df['condition'] == condition]
        means = cond_df.groupby('actual_context')['perplexity'].mean().sort_index()
        hl = compute_half_life(means.index.values, means.values, 0.5)
        row[condition] = hl
    
    row['ratio'] = row['shuffled'] / row['intact'] if row['intact'] > 0 else np.nan
    passage_half_lives.append(row)
    print(f"  {passage:<12} intact={row['intact']:>5.0f}  shuffled={row['shuffled']:>5.0f}  ratio={row['ratio']:.2f}x")

hl_df = pd.DataFrame(passage_half_lives)
print(f"\n  Mean ratio: {hl_df['ratio'].mean():.2f}x (±{hl_df['ratio'].std():.2f})")
```

```python
# Step-by-step benefit analysis
print("\n" + "=" * 70)
print("STEP-BY-STEP BENEFIT (how much does each segment of context add?)")
print("=" * 70)

# Define context segments
segments = [(4, 16), (16, 32), (32, 64), (64, 128), (128, 256), (256, 384)]

for condition in ['intact', 'shuffled']:
    cond_df = df[df['condition'] == condition]
    means = cond_df.groupby('actual_context')['perplexity'].mean()
    
    print(f"\n{condition.upper()}:")
    
    for start, end in segments:
        # Find closest actual context values
        available = means.index.values
        if len(available) == 0:
            continue
        start_ctx = available[np.argmin(np.abs(available - start))]
        end_ctx = available[np.argmin(np.abs(available - end))]
        
        if start_ctx in means.index and end_ctx in means.index and start_ctx != end_ctx:
            ppl_start = means[start_ctx]
            ppl_end = means[end_ctx]
            reduction = ppl_start - ppl_end
            print(f"  {start_ctx:3d}→{end_ctx:3d} tokens: {ppl_start:.1f}→{ppl_end:.1f} (Δ={reduction:+.2f})")
```

```python
# Statistical comparison at key context lengths
from scipy import stats

print("\n" + "=" * 70)
print("STATISTICAL COMPARISON: Intact vs Shuffled")
print("=" * 70)

# Test at key points
key_contexts = [16, 32, 64, 128, 256, 384]

for target_ctx in key_contexts:
    # Find closest actual context
    available = df['actual_context'].unique()
    if len(available) == 0:
        continue
    ctx = available[np.argmin(np.abs(available - target_ctx))]
    
    intact_vals = df[(df['condition']=='intact') & (df['actual_context']==ctx)]['perplexity']
    shuffled_vals = df[(df['condition']=='shuffled') & (df['actual_context']==ctx)]['perplexity']
    
    if len(intact_vals) >= 2 and len(shuffled_vals) >= 2:
        t, p = stats.ttest_ind(intact_vals, shuffled_vals)
        diff = shuffled_vals.mean() - intact_vals.mean()
        sig = "***" if p < 0.001 else "**" if p < 0.01 else "*" if p < 0.05 else ""
        print(f"Context ~{ctx:3d}: intact={intact_vals.mean():.2f}, shuffled={shuffled_vals.mean():.2f}, diff={diff:+.2f}, p={p:.4f} {sig}")
```
